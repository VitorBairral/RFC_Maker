```python
import numpy as np
from scipy.special import jv
import scipy.constants as sc 
from scipy.integrate import quad
```

```python
#Parameters
n = np.sqrt(sc.mu_0 / sc.epsilon_0)
Vc = 0.5*10e5
d = 0.3
R = 0.2296
w0 = 2 * np.pi * 5*10e7
mu = sc.mu_0 * 0.999995
sigma = 5.8 * 10e6
delta = np.sqrt(2/(sigma * mu * w0))
Rs = 1 / (sigma * delta)
#Rs = 20 * 10e-10
Tcav = d / sc.c

T = np.sin((w0*d)/(2*sc.c))/((w0*d)/(2*sc.c))
E0 = Vc / (d*T)  
Hpk = (E0/n) * jv(1, 1.84) / 79.6
print(T)
```

```python
# Figures of Merit
#Energy
U = ((np.pi * sc.epsilon_0 * E0**2)/2) * jv(1, 2.405)**2 * d * R**2
#Geometric factor:
G = n * ((2.405*d)/(2*(d+R)))
#Quality factor
Q0 = G / Rs 
#Dissipated power
Pc = (w0 * U) / Q0
Eacc = Vc / d
Epk = E0
Ra = Vc**2 / Pc

print(f" G: {G} Ohm\n Ra/Q0: {Ra/Q0} Ohm/cell\n Epk/Eacc: {Epk/Eacc} \n Hpk/Eacc: {Hpk/Eacc} Oe/(V/m)")
```
